## 4. Code Implementation

Python and Pandas were used to explore the dataset. The code checks the shape and data types, separates numerical and categorical columns, counts missing values and duplicates, and prepares a summary of the statistics and the missing-data percentage.

**Code and Output (Structure and data types):**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("housing_uncleaned.csv")
```

**Code and Output (Statistics and missing percentage):**

```python
df.describe()
```

```python
quality_summary = pd.DataFrame({
      "Missing Values": df.isnull().sum(),
          "Missing %": (df.isnull().sum() / len(df) * 100).round(2)
          })

quality_summary[quality_summary["Missing Values"] > 0].sort_values(
              "Missing %", ascending=False
              )
```

## 5. Screenshots & Error Handling

The checks showed several data quality issues in the raw dataset. These were recorded in the table below and will be corrected in Week 3.

**Data Quality Issues Identified:**

| Issue | Column(s) | Observation | Planned Handling |
|---|---|---|---|
| Missing values | 16 of 18 columns (all except date and country) | 1,219 missing cells; 79 to 107 per numerical column | Median / mode imputation |
| Duplicate records | All columns | 75 duplicate rows (1.6%) | Remove duplicates |
| Invalid zero values | price, bedrooms, bathrooms | Minimum value is 0 | Review and filter |
| Extreme price | price | Maximum 26.59M against a median of about 465K | Check as outlier |
| Date stored as text | date | Data type is object | Convert to datetime |

**Key Insights:**

- Missing data is low in every column (at most 2.29%), so no column has to be dropped.
- Price is right-skewed: the mean (554,575) is higher than the median (464,900).
- Less than 1% of the houses are on the waterfront, and the median of yr_renovated is 0, which means most houses were never renovated.
- Zero values in price, bedrooms and bathrooms are invalid and need treatment.
