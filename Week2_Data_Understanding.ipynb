**Week** **2**: **Data** **Understanding**

**Objective**:

Understand the size, structure, features, data types, missing values, duplicate records and overall quality of the housing dataset.

### Setup

**Purpose:** Import the libraries and load the raw dataset so this notebook runs on its own.

_Place `housing_uncleaned.csv` in the same folder (or upload it in Colab) before running._

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("housing_uncleaned.csv")
```

### Check Shape and Column Names

**Purpose:** Find the number of rows and columns and list all available features.

```python
print("Shape:", df.shape)

print("\nColumns:", df.columns.tolist())
```

**Insight:** The dataset has **4,675 rows and 18 columns** (price, bedrooms, bathrooms, sqft values, floors, waterfront, view, condition, year built/renovated, street, city, statezip, country).

### Check Dataset Information

**Purpose:** Understand the data types and the non-null count of every column.

```python
df.info()
```

**Insight:** 13 columns are float and 5 are object. `date` is stored as text and should be converted to datetime. Most numeric columns have only about 4,570 to 4,596 non-null values, so each one has some missing data.

### Identify Feature Types

**Purpose:** Separate numerical and categorical columns for later analysis.

```python
num_cols = df.select_dtypes(include=np.number).columns.tolist()
cat_cols = df.select_dtypes(exclude=np.number).columns.tolist()

print("Numerical columns:", num_cols)

print("\nCategorical columns:", cat_cols)
```

**Insight:** There are **13 numerical** and **5 categorical** columns (date, street, city, statezip, country). `waterfront`, `view` and `condition` are stored as numbers but behave like categories.

### Check Missing Values

**Purpose:** Count the missing values in every column.

```python
df.isnull().sum()
```

**Insight:** Every column except `date` and `country` has missing values: about 79 to 107 in each numeric column and only 8 to 9 in `street`, `city` and `statezip`. The total is **1,219 missing cells**.

### Check Duplicate Records

**Purpose:** Count fully duplicated rows.

```python
print("Duplicate rows:", df.duplicated().sum())
```

**Insight:** There are **75 duplicate rows (about 1.6%)**. Keeping them would count the same house twice and distort averages and correlations.

### Check Unique Values of Categorical Columns

**Purpose:** List the number of unique values and the values themselves for each categorical column.

```python
for col in cat_cols:
      print(col, "->", df[col].nunique(), "unique values")
      print(df[col].unique(), "\n")
```

**Insight:** `date` has 70 unique values (sales fall in May to July 2014). `country` shows USA in every row seen, so it carries almost no information. `street` is nearly unique per house and is not useful for modelling.

### Statistical Summary

**Purpose:** Understand the basic statistics (mean, spread, min, max) of the numerical features.

```python
df.describe()
```

**Insight:** Price mean (about 554K) is higher than the median (about 465K), so prices are **right-skewed**. The maximum price is **26.59M** and the minimum is **0**, which is invalid. `bedrooms` and `bathrooms` also have minimum 0. Less than 1% of houses are waterfront (mean 0.007). `yr_renovated` median is 0, which means most houses were never renovated.

### Data Quality Summary

**Purpose:** Show the missing percentage of every column that has missing data, sorted from highest to lowest.

```python
quality_summary = pd.DataFrame({
      "Missing Values": df.isnull().sum(),
          "Missing %": (df.isnull().sum() / len(df) * 100).round(2)
          })

quality_summary[quality_summary["Missing Values"] > 0].sort_values(
              "Missing %", ascending=False
              )
```

**Insight:** Missing data is small everywhere: the highest is `sqft_living` with 2.29%, then `floors` 2.25% and `yr_built` 2.12%. Since no column crosses 2.3%, filling values with median / mode is safe and no column needs to be dropped.

### Week 2 Conclusion

The raw dataset has 4,675 rows and 18 columns (13 numerical and 5 categorical). It contains 1,219 missing values (no column above 2.3%) and 75 duplicate records. The summary statistics show right-skewed prices and invalid zero values in price, bedrooms and bathrooms.
